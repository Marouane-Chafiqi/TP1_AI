```python
import nltk, spacy
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
import re

text = "Bonjour !!! Je M'appelle Marouane Chafiqi, étudiant en master TEE, et j'adore le NLP...@Developpeur !"

text_lower = text.lower()
text_clean = re.sub(r"[^\w\s]", " ", text_lower)
text_clean = re.sub(r"\s+", " ", text_clean).strip()

print("Avant :", text)
print("Après :", text_clean)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens = word_tokenize(text_clean, language='french')
print("Tokens :", tokens)

stop_fr = set(stopwords.words('french'))
print("Stopwords :", stop_fr)

stop_fr.add("bonjour")

filtered = []
for mot in tokens:
    if mot not in stop_fr:
        filtered.append(mot)
print("Mots restants :", filtered)

pourcentage = (len(tokens) - len(filtered)) / len(tokens) * 100
print("Pourcentage supprimé :", pourcentage, "%")
```

```python
import spacy

nlp = spacy.load("fr_core_news_sm")

doc = nlp(" ".join(filtered))

for mot in doc:
    print(mot.text, "->", mot.lemma_)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

lemmas = [mot.lemma_ for mot in doc]

corpus = [
    " ".join(lemmas),
    "j aime apprendre le NLP",
    "le NLP est génial"
]
print(corpus)
```

```python
bow = CountVectorizer()
matrice_bow = bow.fit_transform(corpus)

tableau_bow = pd.DataFrame(matrice_bow.toarray(), columns=bow.get_feature_names_out())
print(tableau_bow)
```

```python
tfidf = TfidfVectorizer()
matrice_tfidf = tfidf.fit_transform(corpus)

tableau_tfidf = pd.DataFrame(matrice_tfidf.toarray(), columns=tfidf.get_feature_names_out())
print(tableau_tfidf.round(2))
```

```python
for i in range(len(corpus)):
    ligne = tableau_tfidf.iloc[i]
    print("phrase", i + 1, "-> mot important :", ligne.idxmax())
```
